```sql
SET use_cached_result = false;

SELECT station_name AS start_station_name, total_hires_started AS total_hires
FROM workspace.tfl_lakehouse.gold_station_activity
ORDER BY total_hires DESC
LIMIT 10;
-- Slowest query: ~7s, 10.26MB, 1 file read, 0 pruned.
-- Optimised by querying Gold table: 0.4s, 51.86KB, 1 file read, 0 pruned. 

SET use_cached_result = false;
SELECT date_trunc('month', start_timestamp) AS month, COUNT(*) AS total_hires
FROM workspace.tfl_lakehouse.silver
GROUP BY date_trunc('month', start_timestamp)
ORDER BY month;

SET use_cached_result = false;
SELECT date_trunc('month', start_timestamp) AS month,
       COUNT(*) AS hires,
       ROUND(100 * AVG(CASE WHEN bike_model = 'PBSC_EBIKE' THEN 1.0 ELSE 0.0 END), 1) AS pct_ebike
FROM workspace.tfl_lakehouse.silver
GROUP BY date_trunc('month', start_timestamp)
ORDER BY month;

SET use_cached_result = false;
SELECT bike_number, bike_model, COUNT(*) AS rides,
       ROUND(AVG(total_duration_ms) / 60000, 1) AS avg_ride_min
FROM workspace.tfl_lakehouse.silver
GROUP BY bike_number, bike_model
ORDER BY rides DESC
LIMIT 10;

```
