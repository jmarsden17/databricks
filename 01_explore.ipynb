```python
data = spark.read.format("csv").option("header", True).load("/Volumes/workspace/tfl/raw/tfl_cycling_2026/439JourneyDataExtract01Mar2026-15Mar2026.csv")

```

```python
data.count()
```

```python
from pyspark.sql import functions as F

null_counts = data.select(
    [F.sum(F.col(c).isNull().cast("int")).alias(c) for c in data.columns]
)
display(null_counts)
```

```python
data.groupBy(F.col("Number")).agg(F.count("*").alias("count")).filter(F.col("count") > 1).show()
```

```python
data.orderBy((F.col("Start date"))).show(5)
```

```python
data = data.withColumn("Total duration (ms)", F.col("Total duration (ms)").cast("bigint"))

data.orderBy(F.col("Total duration (ms)")).show()
```

```python
data.orderBy(F.col("Total duration (ms)"),ascending = False).show()
```

```python
hires_per_day = (
    data.withColumn("day", F.to_date(F.col("Start date"), "yyyy-MM-dd HH:mm"))
    .groupBy("day")
    .agg(F.count("*").alias("hires"))
)
hires_per_day.display()
```
